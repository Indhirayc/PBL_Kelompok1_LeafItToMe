```python
import tensorflow as tf

dataset_path = '/content/drive/MyDrive/dataset'
batch_size = 32
img_size = (244, 244)

# Memuat dataset gambar dari direktori
train_ds = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2,#20% untuk validasi
    subset="training",
    seed=123, #Digunakan untuk pemisah antar data agar konsisten
    image_size=img_size,
    batch_size=batch_size
)

# Memuat data Validasi
val_ds = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2,
    subset="validation",
    seed = 123,
    image_size=img_size,
    batch_size=batch_size
)

#Layer Normalisasi (mengubah piksel dari 1-255 menjadi 0-1)
normalization_layer = tf.keras.layers.Rescaling(1./255)

#Menambahkan layer normalisasi ke dataset
train_ds = train_ds.map(lambda x, y: (normalization_layer(x), y))
val_ds = val_ds.map(lambda x, y: (normalization_layer(x), y))

#optimasi peforma
AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.cache().prefetch(buffer_size=AUTOTUNE)
val_ds = val_ds.cache().prefetch(buffer_size=AUTOTUNE)

print("Preprocessing Selesai. Dataset Siap dilatih!")
```

```python
import os
import shutil
import random
from pathlib import Path

# Path dataset dari notebook sebelumnya
SOURCE_DIR = Path(dataset_path)

# Folder untuk menyimpan hasil pembagian
OUTPUT_DIR = Path("/content/drive/MyDrive/dataset")

# Persentase pembagian
TRAIN_RATIO = 0.80
VAL_RATIO = 0.10
TEST_RATIO = 0.10

# Agar hasil pembagian konsisten
random.seed(42)

# Ekstensi file gambar
IMAGE_EXTENSIONS = {
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp"
}

print("Dataset :", SOURCE_DIR)
print("Output  :", OUTPUT_DIR)
```

```python
classes = sorted([
    folder.name
    for folder in SOURCE_DIR.iterdir()
    if folder.is_dir()
    and folder.name not in ["train", "validation", "test"]
])

print("Kelas dataset:")
for i, name in enumerate(classes, 1):
    print(i, name)
```

```python
for split in ["train", "validation", "test"]:
    for class_name in classes:
        folder = OUTPUT_DIR / split / class_name
        folder.mkdir(parents=True, exist_ok=True)

print("Folder dataset berhasil dibuat.")
```

```python
for name in classes:
    folder = SOURCE_DIR / name

    images = [
        f for f in folder.iterdir()
        if f.is_file() and f.suffix.lower() in IMAGE_EXTENSIONS
    ]

    random.shuffle(images)

    total = len(images)

    train_end = int(total * 0.80)

    validation_end = int(total * 0.90)

    train = images[:train_end]
    validation = images[train_end:validation_end]

    test = images[validation_end:]

    print(
        f"{name}: "
        f"total={total}, "
        f"train={len(train)}, "
        f"validation={len(validation)}, "
        f"test={len(test)}"
    )

    for image in train:
        shutil.copy2(
            image,
            OUTPUT_DIR / "train" / name / image.name
        )

    for image in validation:
        shutil.copy2(
            image,
            OUTPUT_DIR / "validation" / name / image.name
        )

    for image in test:
        shutil.copy2(
            image,
            OUTPUT_DIR / "test" / name / image.name
        )

print("Pembagian dataset selesai.")
```

```python
for split in ["train", "validation", "test"]:
    total = 0

    for name in classes:
        folder = OUTPUT_DIR / split / name

        jumlah = len([
            f for f in folder.iterdir()
            if f.is_file() and f.suffix.lower() in IMAGE_EXTENSIONS
        ])

        total += jumlah

    print(f"{split}: {total} gambar")
```

## Pembuatan dan pelatihan model

```python
# Mengimport library
import os
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

print("TensorFlow version:", tf.__version__)
```

```python
from google.colab import drive

drive.mount('/content/gdrive')
```

```python
import os

DATASET_DIR = '/content/gdrive/MyDrive/dataset'

TRAIN_DIR = os.path.join(DATASET_DIR, 'train')
VAL_DIR = os.path.join(DATASET_DIR, 'validation')
TEST_DIR = os.path.join(DATASET_DIR, 'test')

for folder in [TRAIN_DIR, VAL_DIR, TEST_DIR]:
    if not os.path.exists(folder):
        raise FileNotFoundError(f"Folder tidak ditemukan: {folder}")

print("Lokasi training   :", TRAIN_DIR)
print("Lokasi validation :", VAL_DIR)
print("Lokasi testing    :", TEST_DIR)
```

```python
# Menentukan konfigurasi dataset
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42
```

```python
# Memuat data training
train_ds = tf.keras.utils.image_dataset_from_directory(
    TRAIN_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='int',
    shuffle=True,
    seed=SEED
)
```

```python
# Memuat data validation
val_ds = tf.keras.utils.image_dataset_from_directory(
    VAL_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='int',
    shuffle=False
)
```

```python
# Memuat data testing
test_ds = tf.keras.utils.image_dataset_from_directory(
    TEST_DIR,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='int',
    shuffle=False
)
```

```python
# Memeriksa nama kelas
class_names = train_ds.class_names
num_classes = len(class_names)

print("Nama kelas:", class_names)
print("Jumlah kelas:", num_classes)
```

```python
# Mengoptimalkan pemuatan data
AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(buffer_size=AUTOTUNE)
val_ds = val_ds.prefetch(buffer_size=AUTOTUNE)
test_ds = test_ds.prefetch(buffer_size=AUTOTUNE)

print("Dataset siap digunakan.")
```

```python
# Membuat data augmentation
data_augmentation = tf.keras.Sequential([
    layers.RandomFlip('horizontal'),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
    layers.RandomContrast(0.1)
], name='data_augmentation')

print("Data augmentation berhasil dibuat.")
```

### A. Model CNN

```python
# Membuat arsitektur CNN
cnn_model = models.Sequential([
    layers.Input(shape=(224, 224, 3)),

    # Augmentasi data
    data_augmentation,

    # Normalisasi piksel dari 0–255 menjadi 0–1
    layers.Rescaling(1./255),

    # Ekstraksi fitur
    layers.Conv2D(
        32,
        kernel_size=3,
        activation='relu'
    ),
    layers.MaxPooling2D(),

    layers.Conv2D(
        64,
        kernel_size=3,
        activation='relu'
    ),
    layers.MaxPooling2D(),

    layers.Conv2D(
        128,
        kernel_size=3,
        activation='relu'
    ),
    layers.MaxPooling2D(),

    layers.Conv2D(
        256,
        kernel_size=3,
        activation='relu'
    ),
    layers.MaxPooling2D(),

    # Klasifikasi
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.3),

    layers.Dense(
        128,
        activation='relu'
    ),
    layers.Dropout(0.3),

    layers.Dense(
        num_classes,
        activation='softmax'
    )
], name='LeafItToMe_CNN')
```

```python
# Menampilkan struktur CNN
cnn_model.summary()
```

```python
# Compile model CNN
cnn_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("Model CNN berhasil di-compile.")
```

```python
# Membuat callback CNN
cnn_callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor='val_loss',
        patience=3,
        restore_best_weights=True
    ),

    tf.keras.callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=2,
        min_lr=1e-6
    )
]
```

```python
# Melatih model CNN
import os
import json
from tensorflow.keras.callbacks import CSVLogger

CNN_EPOCHS = 15

# Lokasi dataset dan model
DATASET_DIR = "/content/gdrive/MyDrive/dataset"
TRAIN_DIR = os.path.join(DATASET_DIR, "train")
MODEL_DIR = "/content/gdrive/MyDrive/LeafItToMe/models"

os.makedirs(MODEL_DIR, exist_ok=True)

# Mengambil nama kelas langsung dari folder train
class_names = sorted([
    folder_name
    for folder_name in os.listdir(TRAIN_DIR)
    if os.path.isdir(os.path.join(TRAIN_DIR, folder_name))
])

dataset_num_classes = len(class_names)
model_num_classes = cnn_model.output_shape[-1]

print("Nama kelas       :", class_names)
print("Jumlah kelas     :", dataset_num_classes)
print("Output model CNN :", model_num_classes)

# Memastikan output model sesuai jumlah kelas
if model_num_classes != dataset_num_classes:
    raise ValueError(
        f"Jumlah kelas tidak cocok: dataset memiliki "
        f"{dataset_num_classes} kelas, tetapi output CNN memiliki "
        f"{model_num_classes} kelas. Jalankan ulang sel arsitektur CNN."
    )

# Lokasi penyimpanan riwayat training
CNN_CSV_PATH = os.path.join(
    MODEL_DIR,
    "history_cnn.csv"
)

CNN_JSON_PATH = os.path.join(
    MODEL_DIR,
    "history_cnn.json"
)

# CSVLogger menyimpan hasil setiap epoch secara otomatis
cnn_csv_logger = CSVLogger(
    CNN_CSV_PATH,
    append=False
)

# Menggabungkan callback lama dengan CSVLogger
all_cnn_callbacks = list(cnn_callbacks) + [cnn_csv_logger]

# Melatih model CNN
history_cnn = cnn_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=CNN_EPOCHS,
    callbacks=all_cnn_callbacks
)

# Menyimpan seluruh history ke JSON setelah training selesai
with open(CNN_JSON_PATH, "w") as history_file:
    json.dump(history_cnn.history, history_file)

print("\nPelatihan CNN selesai.")
print("History CSV :", CNN_CSV_PATH)
print("History JSON:", CNN_JSON_PATH)
```

```python
# Menyimpan model CNN
MODEL_DIR = '/content/drive/MyDrive/LeafItToMe/models'

os.makedirs(
    MODEL_DIR,
    exist_ok=True
)

CNN_MODEL_PATH = os.path.join(
    MODEL_DIR,
    'leafittome_cnn.keras'
)

cnn_model.save(CNN_MODEL_PATH)

print("Model CNN berhasil disimpan.")
print("Lokasi:", CNN_MODEL_PATH)
```

### B. Model MobileNetV2

```python
# Memuat MobileNetV2
base_model = MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights='imagenet'
)

# Membekukan layer bawaan MobileNetV2
base_model.trainable = False

print("MobileNetV2 berhasil dimuat.")
print("Jumlah layer:", len(base_model.layers))
```

```python
# Membuat model MobileNetV2
inputs = layers.Input(
    shape=(224, 224, 3)
)

# Data augmentation
x = data_augmentation(inputs)

# Preprocessing khusus MobileNetV2
x = preprocess_input(x)

# Ekstraksi fitur
x = base_model(
    x,
    training=False
)

# Klasifikasi
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)

outputs = layers.Dense(
    num_classes,
    activation='softmax'
)(x)

mobilenet_model = tf.keras.Model(
    inputs=inputs,
    outputs=outputs,
    name='LeafItToMe_MobileNetV2'
)
```

```python
# Menampilkan struktur MobileNetV2
mobilenet_model.summary()
```

```python
# Compile MobileNetV2
mobilenet_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("MobileNetV2 berhasil di-compile.")
```

```python
# Membuat callback MobileNetV2
mobilenet_callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor='val_loss',
        patience=3,
        restore_best_weights=True
    ),

    tf.keras.callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=2,
        min_lr=1e-6
    )
]
```

```python
# Melatih model MobileNetV2
import os
import json
from tensorflow.keras.callbacks import CSVLogger

MOBILENET_EPOCHS = 15

# Lokasi dataset dan model
DATASET_DIR = "/content/gdrive/MyDrive/dataset"
TRAIN_DIR = os.path.join(DATASET_DIR, "train")
MODEL_DIR = "/content/gdrive/MyDrive/LeafItToMe/models"

os.makedirs(MODEL_DIR, exist_ok=True)

# Mengambil nama kelas langsung dari folder train
class_names = sorted([
    folder_name
    for folder_name in os.listdir(TRAIN_DIR)
    if os.path.isdir(os.path.join(TRAIN_DIR, folder_name))
])

dataset_num_classes = len(class_names)
model_num_classes = mobilenet_model.output_shape[-1]

print("Nama kelas              :", class_names)
print("Jumlah kelas            :", dataset_num_classes)
print("Output model MobileNetV2:", model_num_classes)

# Memastikan output model sesuai jumlah kelas
if model_num_classes != dataset_num_classes:
    raise ValueError(
        f"Jumlah kelas tidak cocok: dataset memiliki "
        f"{dataset_num_classes} kelas, tetapi output MobileNetV2 memiliki "
        f"{model_num_classes} kelas. "
        "Jalankan ulang sel arsitektur MobileNetV2."
    )

# Lokasi penyimpanan riwayat training
MOBILENET_CSV_PATH = os.path.join(
    MODEL_DIR,
    "history_mobilenetv2.csv"
)

MOBILENET_JSON_PATH = os.path.join(
    MODEL_DIR,
    "history_mobilenetv2.json"
)

# Menyimpan hasil setiap epoch secara otomatis
mobilenet_csv_logger = CSVLogger(
    MOBILENET_CSV_PATH,
    append=False
)

# Menggabungkan callback lama dengan CSVLogger
all_mobilenet_callbacks = (
    list(mobilenet_callbacks) + [mobilenet_csv_logger]
)

# Melatih model MobileNetV2
history_mobilenet = mobilenet_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=MOBILENET_EPOCHS,
    callbacks=all_mobilenet_callbacks
)

# Menyimpan seluruh history ke JSON setelah training selesai
with open(MOBILENET_JSON_PATH, "w") as history_file:
    json.dump(history_mobilenet.history, history_file)

print("\nPelatihan MobileNetV2 selesai.")
print("History CSV :", MOBILENET_CSV_PATH)
print("History JSON:", MOBILENET_JSON_PATH)
```

```python
# Membuat fungsi grafik
def plot_training_history(history, model_name):
    training_accuracy = history.history['accuracy']
    validation_accuracy = history.history['val_accuracy']

    training_loss = history.history['loss']
    validation_loss = history.history['val_loss']

    epochs = range(
        1,
        len(training_accuracy) + 1
    )

    plt.figure(figsize=(12, 4))

    # Grafik accuracy
    plt.subplot(1, 2, 1)

    plt.plot(
        epochs,
        training_accuracy,
        label='Training Accuracy'
    )

    plt.plot(
        epochs,
        validation_accuracy,
        label='Validation Accuracy'
    )

    plt.title(f'Accuracy {model_name}')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.grid(alpha=0.3)

    # Grafik loss
    plt.subplot(1, 2, 2)

    plt.plot(
        epochs,
        training_loss,
        label='Training Loss'
    )

    plt.plot(
        epochs,
        validation_loss,
        label='Validation Loss'
    )

    plt.title(f'Loss {model_name}')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(alpha=0.3)

    plt.tight_layout()
    plt.show()
```

```python
import os
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    f1_score,
    precision_score,
    recall_score,
)

DATASET_DIR = Path('/content/gdrive/MyDrive/dataset')
MODEL_DIR = Path('/content/gdrive/MyDrive/LeafItToMe/models')
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

print('Folder dataset:', DATASET_DIR)
print('Folder model  :', MODEL_DIR)
```

Memuat Dataset

```python
from google.colab import drive

drive.mount('/content/gdrive', force_remount=True)

split_dirs = {
    'train': DATASET_DIR / 'train',
    'validation': DATASET_DIR / 'validation',
    'test': DATASET_DIR / 'test',
}

missing_dirs = [name for name, path in split_dirs.items() if not path.exists()]
if missing_dirs:
    raise FileNotFoundError(
        'Folder dataset belum ditemukan: '
        + ', '.join(missing_dirs)
        + '. Pastikan shortcut folder dataset sudah tersedia di My Drive.'
    )

train_ds = tf.keras.utils.image_dataset_from_directory(
    split_dirs['train'],
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='int',
    shuffle=False,
    seed=SEED,
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    split_dirs['validation'],
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='int',
    shuffle=False,
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    split_dirs['test'],
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode='int',
    shuffle=False,
)

class_names = train_ds.class_names
print('Nama kelas:', class_names)
print('Jumlah kelas:', len(class_names))
```

Memuat Model

```python
model_filenames = {
    'CNN': 'leafittome_cnn.keras',
    'MobileNetV2': 'leafittome_mobilenetv2.keras',
}

candidate_model_dirs = [
    MODEL_DIR,
    Path('/content/gdrive/MyDrive/dataset/models'),
    Path('/content/gdrive/MyDrive/models'),
    Path('/content/drive/MyDrive/dataset/models'),
]

model_paths = {}
for model_name, filename in model_filenames.items():
    matching_paths = [
        directory / filename
        for directory in candidate_model_dirs
        if (directory / filename).exists()
    ]

    if matching_paths:
        model_paths[model_name] = matching_paths[0]

missing_models = [
    name for name in model_filenames
    if name not in model_paths
]

if missing_models:
    print('Folder model yang diperiksa:')
    for directory in candidate_model_dirs:
        print(' -', directory)

    saved_models = list(Path('/content/gdrive/MyDrive').rglob('*.keras'))
    if saved_models:
        print('\nFile .keras yang ditemukan:')
        for path in saved_models:
            print(' -', path)
    else:
        print('\nTidak ada file .keras di Google Drive.')

    raise FileNotFoundError(
        'Model belum ditemukan: '
        + ', '.join(missing_models)
        + '. Pastikan cell penyimpanan model pada tahap 04 sudah dijalankan '
        + 'dan file .keras tersimpan di Google Drive.'
    )

models_to_evaluate = {
    name: tf.keras.models.load_model(path)
    for name, path in model_paths.items()
}

for name, path in model_paths.items():
    print(f'{name}: {path}')
    models_to_evaluate[name].summary()
```

Matrix Evaluasi

```python
def evaluate_model(model_name, model, dataset):
    loss, accuracy = model.evaluate(dataset, verbose=1)

    y_true = np.concatenate([labels.numpy() for _, labels in dataset])
    probabilities = model.predict(dataset, verbose=0)
    y_pred = np.argmax(probabilities, axis=1)

    precision = precision_score(
        y_true,
        y_pred,
        average='weighted',
        zero_division=0,
    )
    recall = recall_score(
        y_true,
        y_pred,
        average='weighted',
        zero_division=0,
    )
    f1 = f1_score(
        y_true,
        y_pred,
        average='weighted',
        zero_division=0,
    )

    print(f'\nHasil evaluasi {model_name}')
    print(f'Loss      : {loss:.4f}')
    print(f'Accuracy  : {accuracy:.4f}')
    print(f'Precision : {precision:.4f}')
    print(f'Recall    : {recall:.4f}')
    print(f'F1-score  : {f1:.4f}')
    print('\nClassification report:')
    print(classification_report(
        y_true,
        y_pred,
        target_names=class_names,
        zero_division=0,
    ))

    return {
        'model': model_name,
        'loss': loss,
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'f1_score': f1,
        'y_true': y_true,
        'y_pred': y_pred,
    }

results = {
    name: evaluate_model(name, model, test_ds)
    for name, model in models_to_evaluate.items()
}
```

Confusion Matrix

```python
for model_name, result in results.items():
    matrix = confusion_matrix(
        result['y_true'],
        result['y_pred'],
    )

    display = ConfusionMatrixDisplay(
        confusion_matrix=matrix,
        display_labels=class_names,
    )
    display.plot(
        xticks_rotation=90,
        cmap='Blues',
        values_format='d',
    )
    plt.title(f'Confusion Matrix - {model_name}')
    plt.tight_layout()
    plt.show()
```

Perbandingan Performa Model

```python
metric_names = ['accuracy', 'precision', 'recall', 'f1_score']
metric_labels = ['Accuracy', 'Precision', 'Recall', 'F1-score']

x = np.arange(len(metric_names))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 5))
for index, (model_name, result) in enumerate(results.items()):
    values = [result[metric] for metric in metric_names]
    offset = (index - 0.5) * width
    bars = ax.bar(x + offset, values, width, label=model_name)
    ax.bar_label(bars, fmt='%.3f', padding=3, fontsize=9)

ax.set_ylim(0, 1.05)
ax.set_ylabel('Nilai')
ax.set_title('Perbandingan Metrik Evaluasi Model')
ax.set_xticks(x, metric_labels)
ax.legend()
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

best_model = max(results, key=lambda name: results[name]['f1_score'])
print(f'Model dengan F1-score tertinggi: {best_model}')
```
